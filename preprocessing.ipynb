```python
#Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
#Load the Dataset

file_path = Path(
    "data/diabetes_012_health_indicators_BRFSS2015.csv"
)

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())

```

```python
#Explore Dataset

# First five records
display(df.head())

# Last five records
display(df.tail())

# Dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Data types
print("\nData types:")
print(df.dtypes)

# General dataset information
df.info()

# Descriptive statistics
display(df.describe().T)
```

```python
#Check Missing Values

missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_summary)

print(
    "Total missing values:",
    df.isnull().sum().sum()
)
```

```python

# 1. Identify duplicate records
duplicate_mask = df.duplicated(keep="first")

duplicate_count = duplicate_mask.sum()

duplicate_percentage = (
    duplicate_count / len(df)
) * 100

print("Total records:", len(df))
print("Duplicate records:", duplicate_count)
print(
    "Duplicate percentage:",
    round(duplicate_percentage, 2), "%"
)

# 2. Display repeated observations
print("\nSample duplicate records:")
display(df[df.duplicated(keep=False)].head(10))


# 3. Calculate duplicates by diabetes class
duplicate_analysis = pd.DataFrame({
    "Total": df["Diabetes_012"].value_counts(),
    "Duplicates": df.loc[
        duplicate_mask, "Diabetes_012"
    ].value_counts()
}).fillna(0)

# 4. Calculate duplicate percentages
duplicate_analysis["Duplicate_Percentage"] = (
    duplicate_analysis["Duplicates"]
    / duplicate_analysis["Total"]
) * 100

duplicate_analysis["Duplicate_Percentage"] = (
    duplicate_analysis["Duplicate_Percentage"].round(2)
)

# 5. Display class-wise analysis
print("\nDuplicate Analysis by Diabetes Status:")
display(duplicate_analysis.sort_index())


# 6. Verify duplicate counts
print("Total duplicate records:", duplicate_count)
print(
    "Duplicates across all classes:",
    int(duplicate_analysis["Duplicates"].sum())
)

# 7. Duplicate handling decision
# Retain identical rows because multiple survey
# respondents may have the same health indicators.

print("\nDecision: Duplicate records retained.")
```

```python

#Identify duplicate records
duplicate_mask = df.duplicated(keep="first")

# Calculate duplicates by diabetes class
duplicate_analysis = pd.DataFrame({
    "Total": df["Diabetes_012"].value_counts(),
    "Duplicates": df.loc[
        duplicate_mask, "Diabetes_012"
    ].value_counts()
}).fillna(0)

# Calculate duplicate percentages
duplicate_analysis["Duplicate_Percentage"] = (
    duplicate_analysis["Duplicates"]
    / duplicate_analysis["Total"]
) * 100

# Round percentages
duplicate_analysis["Duplicate_Percentage"] = (
    duplicate_analysis["Duplicate_Percentage"].round(2)
)

# Display results
display(duplicate_analysis.sort_index())

# Verify total duplicates
print("Total duplicate records:", duplicate_mask.sum())
print(
    "Duplicates across all classes:",
    int(duplicate_analysis["Duplicates"].sum())
)
```

```python
#Correct Data Types

print("Before conversion:")
print(df.dtypes.value_counts())

# Verify no missing, infinite, or fractional values
numeric_values = df.to_numpy(dtype=float)

assert np.isfinite(numeric_values).all()
assert np.equal(
    numeric_values,
    np.floor(numeric_values)
).all()

# Convert whole-number values to integers
df = df.astype("int64")

print("\nAfter conversion:")
print(df.dtypes.value_counts())

display(df.head())
```

```python
#Validate Categorical Values

binary_columns = [
    "HighBP", "HighChol", "CholCheck",
    "Smoker", "Stroke",
    "HeartDiseaseorAttack",
    "PhysActivity", "Fruits", "Veggies",
    "HvyAlcoholConsump", "AnyHealthcare",
    "NoDocbcCost", "DiffWalk", "Sex"
]

for col in binary_columns:
    invalid = (~df[col].isin([0, 1])).sum()
    print(f"{col}: {invalid} invalid values")

valid_ranges = {
    "Diabetes_012": (0, 2),
    "GenHlth": (1, 5),
    "MentHlth": (0, 30),
    "PhysHlth": (0, 30),
    "Age": (1, 13),
    "Education": (1, 6),
    "Income": (1, 8)
}

for col, (minimum, maximum) in valid_ranges.items():
    invalid = (
        ~df[col].between(minimum, maximum)
    ).sum()

    print(f"{col}: {invalid} invalid values")

print(
    "Invalid BMI values:",
    (df["BMI"] <= 0).sum()
)
```

```python
#Outlier Visualisation

plt.figure(figsize=(9, 5))

plt.boxplot(
    df["BMI"].dropna(),
    vert=False
)

plt.title("BMI Outlier Detection")
plt.xlabel("BMI")
plt.grid(axis="x", alpha=0.3)

plt.show()
```

```python

plt.figure(figsize=(9, 5))

plt.hist(
    df["BMI"],
    bins=40
)

plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")

plt.show()
```

```python

#IQR Outlier Detection

Q1 = df["BMI"].quantile(0.25)
Q3 = df["BMI"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["BMI"] < lower_limit) |
    (df["BMI"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

print("Potential outliers:", len(outliers))
```

```python
#Feature Creation

tableau_df = df.copy()

# Create readable diabetes labels
tableau_df["Diabetes_Status"] = (
    tableau_df["Diabetes_012"].map({
        0: "No Diabetes",
        1: "Prediabetes",
        2: "Diabetes"
    })
)

# Create descriptive BMI categories
tableau_df["BMI_Category"] = pd.cut(
    tableau_df["BMI"],
    bins=[0, 18.5, 25, 30, np.inf],
    labels=[
        "Underweight",
        "Normal",
        "Overweight",
        "Obesity"
    ],
    right=False
)

display(
    tableau_df[
        ["Diabetes_012", "Diabetes_Status",
         "BMI", "BMI_Category"]
    ].head(10)
)
```

```python
#Remove Irrelevant Columns

df = df.drop(
    columns=["Unnamed: 0"],
    errors="ignore"
)

tableau_df = tableau_df.drop(
    columns=["Unnamed: 0"],
    errors="ignore"
)

print("ML columns:", len(df.columns))
print("Tableau columns:", len(tableau_df.columns))
```

```python
#Clean Column Names

df.columns = df.columns.str.strip()

tableau_df.columns = (
    tableau_df.columns.str.strip()
)

print(df.columns.tolist())
```

```python
#Final Data Validation

print("FINAL DATASET VALIDATION")

print("\nDataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows retained:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nTarget distribution:")
print(df["Diabetes_012"].value_counts().sort_index())

display(df.head())
display(df.describe().T)

# Final checks
assert df["Diabetes_012"].isin([0, 1, 2]).all()
assert df.notna().all().all()
assert np.isfinite(df.to_numpy(dtype=float)).all()

print("\nFinal validation completed!")
```

```python
#Export Datasets

from pathlib import Path

output_folder = Path("data/processed datasets")
output_folder.mkdir(parents=True, exist_ok=True)

# Export original validated features for ML
df.to_csv(
    output_folder / "diabetes_ml.csv",
    index=False
)

# Export Tableau-ready features and labels
tableau_df.to_csv(
    output_folder / "diabetes_tableau.csv",
    index=False
)

print("Datasets exported successfully!")
print("ML records:", len(df))
print("Tableau records:", len(tableau_df))
```
